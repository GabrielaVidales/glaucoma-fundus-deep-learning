```python
import os
import random
import shutil

# ┅ Rutas base
base_path = "/content/drive/MyDrive/Proyecto_glaucoma"
src_train = os.path.join(base_path, "data_raw/dataset/train")
src_test  = os.path.join(base_path, "data_raw/dataset/test")

dest_base = os.path.join(base_path, "data_Alltrain_Allval_240test")

# ┅ Crear estructura destino
for split in ["train", "val", "test"]:
    for clase in ["normal", "glaucoma"]:
        os.makedirs(os.path.join(dest_base, split, clase), exist_ok=True)

# ┅ Funci%C3%B3n para copiar im%C3%A1genes sin repetir
def copiar_imagenes(origen, destino, cantidad, usadas=set()):
    imagenes = [img for img in os.listdir(origen) if img not in usadas]

    # Check if requested quantity is greater than available images
    if cantidad > len(imagenes):
        print(f"Warning: Not enough images in {origen}. Requested {cantidad}, but only {len(imagenes)} are available. Copying all available.")
        cantidad = len(imagenes)

    seleccion = random.sample(imagenes, cantidad)

    for img in seleccion:
        shutil.copy2(os.path.join(origen, img), os.path.join(destino, img))

    return set(seleccion)

# =========================
# ══ TRAIN y VAL desde train/
# =========================

usadas_normal = set()
usadas_glaucoma = set()

# NORMAL
src_normal = os.path.join(src_train, "normal")
dest_train_normal = os.path.join(dest_base, "train/normal")
dest_val_normal   = os.path.join(dest_base, "val/normal")

#10400, 480
#  train
usadas = copiar_imagenes(src_normal, dest_train_normal, 10400)
usadas_normal.update(usadas)

#  val (distintas)
usadas = copiar_imagenes(src_normal, dest_val_normal, 480, usadas_normal)
usadas_normal.update(usadas)


# GLAUCOMA


src_glaucoma = os.path.join(src_train, "glaucoma")
dest_train_glaucoma = os.path.join(dest_base, "train/glaucoma")
dest_val_glaucoma   = os.path.join(dest_base, "val/glaucoma")

# 6500 , 450
usadas = copiar_imagenes(src_glaucoma, dest_train_glaucoma, 6500)
usadas_glaucoma.update(usadas)

#   val (distintas)
usadas = copiar_imagenes(src_glaucoma, dest_val_glaucoma, 450, usadas_glaucoma)
usadas_glaucoma.update(usadas)


# =========================
# ══ TEST desde test/
# =========================

# NORMAL
#src_test_normal = os.path.join(src_test, "normal")
#dest_test_normal = os.path.join(dest_base, "test/normal")

#copiar_imagenes(src_test_normal, dest_test_normal, 240)

# GLAUCOMA
#src_test_glaucoma = os.path.join(src_test, "glaucoma")
#dest_test_glaucoma = os.path.join(dest_base, "test/glaucoma")

#copiar_imagenes(src_test_glaucoma, dest_test_glaucoma, 240)


#print("✅ Dataset creado correctamente")
```
